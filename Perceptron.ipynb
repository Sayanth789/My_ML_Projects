#Perceptron

A **Perceptron** is the simplest type of neural network — a single neuron that makes a binary decision.

The Perceptron takes inputs x1, x2
Each has weights w1, w2

There's also a bias 'b'

###It computes a weighted sum
z = x1*w1 + z2*w2 + b
This is called linear combination or logit.

It takes a simple threshold function:

if z  > 0:
  output = 1
else:
  output = 0
    


# A minimal Perceptron Implementation (from scratch)



```python
import numpy as np

class Perceptron:
    def __init__(self, learning_rate=0.1, n_iters=1000):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.activation = self._unit_step
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        n_samples, n_features = X.shape

        self.weights = np.zeros(n_features)
        self.bias = 0

        for _ in range(self.n_iters):
            for idx, x_i in enumerate(X):
                linear_output = np.dot(x_i, self.weights) + self.bias
                y_pred = self.activation(linear_output)

                # Update rule
                error = y[idx] - y_pred
                self.weights += self.lr * error * x_i
                self.bias += self.lr * error

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        y_pred = self.activation(linear_output)
        return y_pred

    def _unit_step(self, x):
        return np.where(x >= 0, 1, 0)
```

# Usage

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# AND logic gate labels
y = np.array([0, 0, 0, 1])

p = Perceptron(learning_rate=0.1, n_iters=10)
p.fit(X, y)

print("Weights: ", p.weights)
print("Bias:", p.bias)

print("Predictions:", p.predict(X))
```
