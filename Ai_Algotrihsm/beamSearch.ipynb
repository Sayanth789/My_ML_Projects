```python
import heapq

def beam_search(graph, start, goal, beam_width):
  """ Finds a path from the start to the goal using the beam search algorithm.
  Args:
      graph (dict): A graph represented as a dictionary of nodes.
                    Each node maps to a list of (neighbor, cost) tuples
       start: The starting node.
       goal: The target node
       beam_width (int): The number of candidates to keep at each step
  Returns:
        list: The path from start to goal if found, otherwise None.
  """
  # A  priority queue to store tuples of (cost , path)
  # The smallest cost will have the highest priority
  priority_queue =  [(0, [start])]

  # A set to keep track of visited nodes to avoid cycles
  visited = { start }

  while priority_queue:
    # Get the top candidates (paths) from the priority queue
    candidates = [heapq.heappop(priority_queue) for _ in range(len(priority_queue))]
    heapq.heapify(priority_queue)   # Re-heapify for safety

    next_level_candidates = []
    for cost , path in candidates:
      current_node = path[-1]

      if current_node == goal:
        return path

      # Generate all successor paths for the current path
      for neighbor, neighbor_cost in graph.get(current_node, []):
        new_cost = cost + neighbor_cost
        new_path = path + [neighbor]

        # check for cycles
        if neighbor not in visited:
          visited.add(neighbor)
          next_level_candidates.append((new_cost, new_path))

    # sort all generated candidates and keep only the top beam_width
    next_level_candidates.sort(key=lambda x: x[0])

    for item in next_level_candidates[:beam_width]:
      heapq.heappush(priority_queue, item)


  return None

# Example usage : Finding the shortest path in a weighted graph
if __name__ == "__main__":
  graph = {
      'A': [('B', 3), ('C', 2), ('D', 5)],
      'B': [('E', 4), ('F', 1)],
      'C': [('G', 7)],
      'D': [('F', 3)],
      'E': [('G', 2)],
      'F': [('H', 6)],
      'G': [('H, 1')],
      'H':[]

  }

  start_node = 'A'
  goal_node = 'H'
  beam_width = 2

  path = beam_search(graph, start_node, goal_node, beam_width)

  if path:
      print(f"Path found with beam width {beam_width}: {' -> '.join(path)}")
  else:
      print(f"No path found with beam width {beam_width}.")
```
